```python

import sqlite3

# Conectando ao banco de dados
conexao = sqlite3.connect('banco.db')

# Criando um cursor para executar comandos SQL
cursor = conexao.cursor()

# cursor executa um comando SQL para criar a tabela "produtos" se ela não existir
cursor.execute("""
               
               create table if not exists produtos(
                   
                   id integer primary key autoincrement, 
                   
                   nome text not null,
                   
                   qtd integer not null,
                   
                   preco real not null
                   
               )
               
               
               """)

# commit salva as alterações feitas no banco de dados
conexao.commit()

# fechando a conexão com o banco de dados após terminar as operações
conexao.close()
```
