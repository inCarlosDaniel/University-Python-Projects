```python
import sqlite3  # Importa a biblioteca sqlite3 para trabalhar com bancos de dados SQLite

import os # Importa a biblioteca os para interagir com o sistema operacional, como verificar a existência de arquivos


# Função para criar o banco de dados se ele não existir ou abrir o banco de dados se ele já existir
def criar_database():
    
    
    nome_banco = "banco.db" # Define o nome do arquivo do banco de dados
    
    
    if not os.path.exists(nome_banco): # Verifica se o arquivo do banco de dados não existe
        
        
        print("Banco não encontrado! Criando o banco de dados...") # Imprime uma mensagem indicando que o banco de dados não foi encontrado e que será criado
        
        
    else: # Se o arquivo do banco de dados já existir, imprime uma mensagem indicando que o banco de dados já existe e que será aberto
        
        
        print("Banco de dados ja existe! Abrindo o banco de dados...")
        
criar_database() # Chama a função criar_database para verificar a existência do banco de dados e criar ou abrir conforme necessário
```
