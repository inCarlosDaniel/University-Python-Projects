```python

altura = float(input("Digite a altura: "))  # Solicita ao usuário que insira a altura e converte para float


peso = float(input("Digite o peso: ")) # Solicita ao usuário que insira o peso e converte para float


imc = peso / (altura ** 2)  #Calcula o IMC usando a fórmula peso dividido pela altura ao quadrado


print(f"Seu IMC é: {imc:.2f}") #Imprime o IMC formatado com duas casas decimais



if imc < 18.5:
                                        # Verifica se o IMC é menor que 18.5
    print("Baixo peso")

elif imc < 25:
                                        # Verifica se o IMC é menor que 25
    print("peso normal")

elif imc < 30:
                                        # Verifica se o IMC é menor que 30
    print("Obesidade leve")

elif imc < 35:
                                        # Verifica se o IMC é menor que 35
    print("Obesidade moderada")

else:
                                        #Se o IMC for 35 ou maior, indica obesidade severa.
    print(" Obesidade severa")

```
