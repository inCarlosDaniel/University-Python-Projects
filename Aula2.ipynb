```python
import random  #vai importar a biblioteca random, que tem a função de gerar numeros aleatorios.

N = random.randint(1, 10) # A função randint() é usada para gerar um número inteiro aleatório.

print("Número Sorteado:", N)  # O print é uma função de saida de dados, ele vai mostrar por usuario o resultado.
```

```python
import random  #vai importar a biblioteca random, que tem a função de gerar numeros aleatorios.

alunos = ["Jão", "Eduardo", "Millena", "Carlos", "Pedro"] # Criar uma lista de alunos.

Escolhido = random.choice(alunos) # A função choice() é usada para escolher um elemento aleatório de uma sequência, como uma lista.

print("Aluno Sorteado:", Escolhido)  # O print é uma função de saida de dados, ele vai mostrar por usuario o resultado.
```
