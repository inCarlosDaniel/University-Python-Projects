```python
import math 

#Raiz Quadrada

numero = 49 #Número para calcular a raiz quadrada

raiz = math.sqrt(numero) #Calcula a raiz quadrada usando a função sqrt do módulo math

print(f"A raiz quadrada de {numero}: {raiz}") #Exibe o resultado formatado usando f-string

#saída: a raiz quadrada de 49: 7.0

#Seno de 90 graus (em radianos)

angulo_graus = 90 #Ângulo em graus para calcular o seno

angulo_rad = math.radians(angulo_graus) #Converte o ângulo de graus para radianos usando a função radians do módulo math

seno = math.sin(angulo_rad) #Calcula o seno do ângulo em radianos usando a função sin do módulo math

print(f"O seno de {angulo_graus}°: {seno:.2f}") #Exibe o resultado formatado usando f-string, limitando a saída a 2 casas decimais

#saída: o seno de 90°: 1.0

#Constante Pi

print(f"Pi = {math.pi:.5f}") #Exibe o valor da constante Pi formatado usando f-string, limitando a saída a 5 casas decimais
#saída: pi = 3.14159
```

```python
import math

angulo_graus = float(input("Digite um número de graus: ")) #recebe um número de graus do usuário

angulo_rad = math.radians(angulo_graus) #converte o número de graus para radianos usando a função radians() da biblioteca math

seno = math.sin(angulo_rad) #calcula o seno do ângulo em radianos usando a função sin() da biblioteca math

print(f"O seno de {angulo_graus}°: {seno:.2f}") #imprime o resultado formatado com duas casas decimais usando f-string 

```

```python

import math

#calcula os catetos a = 3 e b = 4, usando math.sqrt para calcular a hipotenusa

a = 3 #cateto a

b = 4 #cateto b

hipotenusa = math.sqrt(a ** 2 + b ** 2 ) #calcula a hipotenusa usando o teorema de Pitágoras

print(f"A hipotenusa do triângulo retângulo com catetos {a} e {b} é: {hipotenusa:.2f}") #imprime o resultado formatado com duas casas decimais 
```

```python
import math

#calcula o logaritmo de 1000 na base 10

numero = 1000 #número pára calcular o logaritmo


logaritmo = math.log(numero, 10) #calcula o logaritmo de 1000 na base 10 usando a função log10() da biblioteca math

print(f"O Logaritmo de {numero} na base 10 é: {logaritmo:.2f}") #imprime o resultado formatado com duas casas decimais usando f-string
```

```python
import math

numero = 4.3

math.ceil(numero) #arredonda o número para cima, retornando o menor inteiro maior ou igual a 4.3, que é 5

math.floor(numero) #arredonda o número para baixo, retornando o maior inteiro menor ou igual a 4.3, que é 4

print(math.ceil(numero),",",math.floor(numero)) #imprime o resultado formatado usando f-string
```

```python
import math

#Calcule o cosseno de 0 graus

angulo_graus = 0 #Ângulo em graus para calcular o cosseno

cosseno = math.cos(math.radians(angulo_graus)) #Calcula o cosseno do ângulo em radianos usando a função cos do módulo math, convertendo o ângulo de graus para radianos com a função radians

print(f"O cosseno de {angulo_graus}°: {cosseno:.2f}") #Exibe o resultado formatado usando f-string, limitando a saída a 2 casas decimais
```

```python
import math

numero = 6 #Número para calcular o fatorial

math.factorial(numero) #Calcula o fatorial de um número inteiro não negativo usando a função factorial do módulo math

print(f"O fatorial de {numero} é: {math.factorial(numero)}") #Exibe o resultado formatado usando f-string
```
