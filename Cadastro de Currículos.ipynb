```python
print("Bem-vindo ao sistema de cadastro de currículos!") # Exibe uma mensagem de boas-vindas para o usuário.

matriz = [] # Cria uma lista vazia chamada "matriz" que será usada para armazenar os currículos cadastrados.

qtdCurriculo = int(input("Quantos currículos você quer cadastrar? ")) # entrada de dados que solicita ao usuário que informe a quantidade de currículos que deseja cadastrar 

for i in range(qtdCurriculo):

    print(f"\nCadastro do {i+1}º Currículo") #saida de dados que exibe uma mensagem indicando o número do currículo que está sendo cadastrado.
                                             #utilizando a variável "i" para contar a quantidade de currículos cadastrados.

    linha = []  # Cria uma lista vazia chamada "linha" que será usada para armazenar as informações de cada currículo individualmente.
    

    linha.append(input("Digite seu Nome: ")) #entrada de dados que solicita ao usuário que informe seu nome.
                                             #armazena o nome na lista "linha" usando o método "append()"

    linha.append(input("Digite seu CPF: ")) #entrada de dados que solicita ao usuário que informe seu CPF
                                            #armazena o CPF na lista "linha" usando o método "append()".


    linha.append(int(input("Digite sua idade: "))) #entrada de dados que solicita ao usuário que informe sua idade, converte a entrada para um número inteiro usando a função "int()" 
                                                   #armazena idade na lista "linha" usando o método "append()".


    linha.append(input("Digite seu telefone: ")) #entrada de dados que solicita ao usuário que informe seu telefone.
                                                 #armazena o telefone na lista "linha" usando o método "append()".


    linha.append(input("Digite seu email: ")) #entrada de dados que solicita ao usuário que informe seu email.
                                              #armazena o email na lista "linha" usando o método "append()".


    linha.append(input("Digite seu estado civil: ")) #entrada de dados que solicita ao usuário que informe seu estado civil.
                                                     #armazena o estado civil na lista "linha" usando o método "append()".


    linha.append(input("Digite seu Endereço: ")) #entrada de dados que solicita ao usuário que informe seu endereço 
                                                 #armazena essa informação na lista "linha" usando o método "append()".


    linha.append(input("Digite seu minicurrículo: ")) #entrada de dados que solicita ao usuário que informe um minicurrículo, um resumo breve de suas experiências.
                                                      #armazena essa informação na lista "linha" usando o método "append()".
                                 

    matriz.append(linha) #adiciona a lista "linha" à lista "matriz" usando o método "append()".


print("\nMATRIZ DE CURRÍCULOS:\n") #saida de dados que exibe uma mensagem indicando que a matriz de currículos será exibida a seguir.

for linha in matriz: #percorre cada elemento da lista "matriz" usando um loop "for" e atribui cada elemento à variável "linha".
    
    print(linha) #saida de dados que exibe o conteúdo de cada "linha" da matriz, as informações de cada currículo cadastrado.
```
