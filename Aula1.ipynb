```python

import matplotlib.pyplot as plt #Fazendo meu primeiro grafico, importando a bibliotecla de grafico, logo apos chamado a bibliotecla no plt.plot e difinido  a cor e difinido o eixo x e o y.

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 18]

plt.plot(x, y, color='purple' , marker='o')
plt.title('Meu Primeiro Gráfico')
plt.ylabel('Eixo X')
plt.ylabel('Eixo Y')
plt.show()
```

```python


Nota1 = float(input("Digite a 1° Nota ")) #recebe a primeiro nota que o usuario digita

Nota2 = float(input("Digite a 2° Nota: ")) #recebe a segundo nota que o usuario digita

Nota3 = float(input("Digite a 3° Nota ")) #recebe a terceira nota que o usuario digita

media = (Nota1 + Nota2 + Nota3) / 3 #Calcula as 3 notas que o usuario digitou e divide por 3

if media >= 6:
                                    #se a media do aluno for maior ou igual a 6, ele vai ta aprovado!
  print("Aluno APROVADO!")

else:
                                    #se caso a media do aluno for menor que 6, ele vai ta reprovado!
  print("Aluno REPROVADO!")

print("A media do Aluno é = ", media)

```
