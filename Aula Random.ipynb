```python
import random

#solteiro cara e coroa

solteiro = ("Cara", "Coroa")   #solteiro receve uma tupla com os valores "Cara" e "Coroa"

solteiro = random.choice(solteiro) #solteiro recebe um valor aleatório da tupla "solteiro"

if solteiro == "Cara":  
                                #se o valor de solteiro for igual a "Cara", então o solteiro é Cara
    print("O solteiro é Cara")
    
else:
                                #se o valor de solteiro for diferente de "Cara", então o solteiro é Coroa
    print("O solteiro é Coroa")
    
    
```

```python
import random

gerar = random.randint(1,20)  #gerar recebe um número inteiro aleatório entre 1 e 20

print("O número gerado foi:", gerar) #imprime o número gerado
```

```python
import random

#Selecione 8 caracteres aleatórios (sem repetição) de uma string de letras.

letras = "JãoPauloSilva"  #letras recebe uma string com todas as letras do alfabeto

selecionar = random.sample(letras, 8) #selecionar recebe uma lista com 8 caracteres aleatórios da string "letras"

print("Os caracteres selecionados foram:", selecionar)  #imprime os caracteres selecionados
```

```python
import random

Playlist = ["Rock", "Pop", "Jazz", "MPB"] #Playlist recebe uma lista com os gêneros musicais "Rock", "Pop", "Jazz" e "MPB"

random.shuffle(Playlist) #embaralha a playlist, ou seja, muda a ordem dos elementos da lista "Playlist" de forma aleatória

print("A playlist embaralhada é:", Playlist) #imprime a playlist embaralhada
```

```python
import random

#Gere um número float entre 0% e 100% com 2 casas decimais.

porcentagem = round(random.random() * 100, 2)  #porcentagem recebe um número float aleatório entre 0 e 1, multiplicado por 100 para obter um número entre 0 e 100, e arredondado para 2 casas decimais

print("A porcentagem gerada foi:", porcentagem, "%") #imprime a porcentagem gerada
```
